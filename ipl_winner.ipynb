```python
def get_player_matches(player_name):
    """
    Returns all matches played by a player in chronological order.
    """
    
    player_data = ball[
        (ball['batter'] == player_name) |
        (ball['bowler'] == player_name)
    ]

    matches_played = (
        player_data[['match_id', 'date']]
        .drop_duplicates()
        .sort_values('date')
        .reset_index(drop=True)
    )

    return matches_played
```

```python
dhoni_matches = get_player_matches("MS Dhoni")

print("Total Matches :", len(dhoni_matches))
dhoni_matches.head(10)
```

```python
def get_last_5_matches(player_name):
    """
    Returns the last 5 matches played by the player.
    """

    matches_played = get_player_matches(player_name)

    return matches_played.tail(5).reset_index(drop=True)
```

```python
# ==========================================================
# PHASE 4 : PLAYER RECENT FORM ENGINE
# STEP 4.1 : LAST 5 MATCH BALL DATA
# ==========================================================

```

```python
def get_last5_ball_data(player_name):

    # Last 5 matches
    last5 = get_last_5_matches(player_name)

    match_ids = last5['match_id'].tolist()

    # Batting OR Bowling records
    player_ball = ball[
        (ball['match_id'].isin(match_ids)) &
        (
            (ball['batter'] == player_name) |
            (ball['bowler'] == player_name)
        )
    ].copy()

    return player_ball
```

```python
dhoni_ball = get_last5_ball_data("V Kohli")

print(dhoni_ball.shape)

dhoni_ball.head()
```

```python
# ==========================================================
# PHASE 4.2 : BATTING SUMMARY (BUG FIX)
# ==========================================================

def get_batting_summary(player_name):

    player_ball = get_last5_ball_data(player_name)

    batting = player_ball[player_ball["batter"] == player_name].copy()

    runs = batting["batsman_runs"].sum()
    balls = len(batting)

    fours = (batting["batsman_runs"] == 4).sum()
    sixes = (batting["batsman_runs"] == 6).sum()

    strike_rate = round((runs / balls) * 100, 2) if balls > 0 else 0

    outs = batting["is_wicket"].sum()
    average = round(runs / outs, 2) if outs > 0 else float(runs)

    # ⭐ BUG FIX
    total_matches = player_ball["match_id"].nunique()

    summary = pd.DataFrame({
        "player": [player_name],
        "matches": [int(total_matches)],
        "batting_runs": [int(runs)],
        "balls_faced": [int(balls)],
        "batting_average": [round(float(average), 2)],
        "strike_rate": [round(float(strike_rate), 2)],
        "fours": [int(fours)],
        "sixes": [int(sixes)]
    })

    return summary
```

```python
# ==========================================================
# PHASE 4.3 : BOWLING SUMMARY (BUG FIX)
# ==========================================================

def get_bowling_summary(player_name):

    player_ball = get_last5_ball_data(player_name)

    bowling = player_ball[player_ball["bowler"] == player_name].copy()

    balls_bowled = len(bowling)
    overs = round(balls_bowled / 6, 1)

    runs_conceded = bowling["total_runs"].sum()
    wickets = bowling["is_wicket"].sum()

    economy = round(runs_conceded / overs, 2) if overs > 0 else 0
    bowling_average = round(runs_conceded / wickets, 2) if wickets > 0 else 0

    # ⭐ BUG FIX
    total_matches = player_ball["match_id"].nunique()

    summary = pd.DataFrame({
        "player": [player_name],
        "matches": [int(total_matches)],
        "balls_bowled": [int(balls_bowled)],
        "overs": [overs],
        "runs_conceded": [int(runs_conceded)],
        "wickets": [int(wickets)],
        "economy": [economy],
        "bowling_average": [bowling_average]
    })

    return summary
```

```python
# ==========================================================
# PHASE 4.4 : PLAYER RECENT FORM (FINAL)
# ==========================================================

def get_player_recent_form(player_name):

    batting = get_batting_summary(player_name)
    bowling = get_bowling_summary(player_name)

    # Merge both summaries
    recent_form = batting.merge(
        bowling.drop(columns=["matches"]),
        on="player",
        how="left"
    )

    return recent_form
```

```python
# ==========================================================
# PHASE 5.2 : TEAM RECENT FORM
# ==========================================================

def get_team_recent_form(team_players):

    team_form = []

    for player in team_players:

        player_form = get_player_recent_form(player)

        team_form.append(player_form)

    team_form = pd.concat(team_form, ignore_index=True)

    return team_form
```

```python
# ==========================================================
# PHASE 5.3 : TEAM BATTING STRENGTH
# ==========================================================

def get_team_batting_strength(team_form):

    batting_strength = pd.DataFrame({

        "total_runs": [team_form["batting_runs"].sum()],

        "average_strike_rate": [
            round(team_form["strike_rate"].mean(), 2)
        ],

        "average_batting_average": [
            round(team_form["batting_average"].mean(), 2)
        ],

        "total_fours": [
            team_form["fours"].sum()
        ],

        "total_sixes": [
            team_form["sixes"].sum()
        ]

    })

    return batting_strength
```

```python
# ==========================================================
# PHASE 5.4 : TEAM BOWLING STRENGTH
# ==========================================================

def get_team_bowling_strength(team_form):

    bowling_strength = pd.DataFrame({

        "total_wickets": [
            team_form["wickets"].sum()
        ],

        "average_economy": [
            round(team_form["economy"].mean(), 2)
        ],

        "average_bowling_average": [
            round(team_form["bowling_average"].mean(), 2)
        ],

        "total_balls_bowled": [
            team_form["balls_bowled"].sum()
        ],

        "total_runs_conceded": [
            team_form["runs_conceded"].sum()
        ]

    })

    return bowling_strength
```

```python
# ==========================================================
# PHASE 5.4.1 : TEAM BOWLING STRENGTH (BOWLERS ONLY)
# ==========================================================

def get_team_bowling_strength(team_form):

    # Only players who bowled
    bowlers = team_form[team_form["balls_bowled"] > 0].copy()

    bowling_strength = pd.DataFrame({

        "bowlers": [len(bowlers)],

        "total_wickets": [
            bowlers["wickets"].sum()
        ],

        "average_economy": [
            round(bowlers["economy"].mean(), 2)
        ],

        "average_bowling_average": [
            round(bowlers["bowling_average"].mean(), 2)
        ],

        "total_balls_bowled": [
            bowlers["balls_bowled"].sum()
        ],

        "total_runs_conceded": [
            bowlers["runs_conceded"].sum()
        ]

    })

    return bowling_strength
```

```python
# ==========================================================
# PHASE 5.5 : OVERALL TEAM STRENGTH
# ==========================================================

def get_team_strength(team_form):

    batting = get_team_batting_strength(team_form)

    bowling = get_team_bowling_strength(team_form)

    team_strength = pd.concat(
        [batting, bowling],
        axis=1
    )

    return team_strength
```

```python
# ============================
# Phase 6 - Match Feature Generator
# ============================

def generate_match_features(team1_players, team2_players):

    # Team 1
    team1_form = get_team_recent_form(team1_players)
    team1_strength = get_team_strength(team1_form)

    # Team 2
    team2_form = get_team_recent_form(team2_players)
    team2_strength = get_team_strength(team2_form)

    # Rename Team 1 columns
    team1_strength = team1_strength.add_prefix("team1_")

    # Rename Team 2 columns
    team2_strength = team2_strength.add_prefix("team2_")

    # Combine into one feature row
    match_features = pd.concat(
        [team1_strength, team2_strength],
        axis=1
    )

    return match_features
```

```python
# ==========================================
# FEATURE ENGINEERING
# INSIGHT 8 - TEAM WIN PERCENTAGE
# ==========================================
```

```python
from collections import defaultdict

# Initialize dictionaries
team_matches = defaultdict(int)
team_wins = defaultdict(int)

# New columns
matches['team1_win_pct'] = 0.0
matches['team2_win_pct'] = 0.0

# Matches must already be sorted by date
for i, row in matches.iterrows():

    team1 = row['team1']
    team2 = row['team2']
    winner = row['winner']

    # Team 1 win %
    if team_matches[team1] == 0:
        matches.at[i, 'team1_win_pct'] = 50.0
    else:
        matches.at[i, 'team1_win_pct'] = round(
            (team_wins[team1] / team_matches[team1]) * 100,
            2
        )

    # Team 2 win %
    if team_matches[team2] == 0:
        matches.at[i, 'team2_win_pct'] = 50.0
    else:
        matches.at[i, 'team2_win_pct'] = round(
            (team_wins[team2] / team_matches[team2]) * 100,
            2
        )

    # Update AFTER current match
    team_matches[team1] += 1
    team_matches[team2] += 1

    if winner == team1:
        team_wins[team1] += 1
    elif winner == team2:
        team_wins[team2] += 1
```

```python
# Create new columns
matches['team1_h2h'] = 0
matches['team2_h2h'] = 0

# Dictionary to store H2H history
h2h_history = {}
```

```python
for i, row in matches.iterrows():

    team1 = row['team1']
    team2 = row['team2']
    winner = row['winner']

    teams = tuple(sorted([team1, team2]))

    if teams not in h2h_history:
        h2h_history[teams] = {team1: 0, team2: 0}

    matches.at[i, 'team1_h2h'] = h2h_history[teams].get(team1, 0)
    matches.at[i, 'team2_h2h'] = h2h_history[teams].get(team2, 0)

    h2h_history[teams][winner] = h2h_history[teams].get(winner, 0) + 1
```

```python
from collections import defaultdict

matches['team1_last5_wins'] = 0
matches['team2_last5_wins'] = 0

team_history = defaultdict(list)
```

```python
for i, row in matches.iterrows():

    team1 = row['team1']
    team2 = row['team2']
    winner = row['winner']

    # Last 5 wins BEFORE current match
    matches.at[i, 'team1_last5_wins'] = sum(team_history[team1][-5:])
    matches.at[i, 'team2_last5_wins'] = sum(team_history[team2][-5:])

    # Update history AFTER current match
    team_history[team1].append(1 if winner == team1 else 0)
    team_history[team2].append(1 if winner == team2 else 0)
```
